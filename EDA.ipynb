## Analysis

### Analysis of each sign for each class

In order to understand those data, i will vizualise the trend of each sign for each categorie of video

```python
def extract_signs(class_means,label, n=875):
    """Extract a list for each sign of the instance

    Args:
        instance (list): instance of data
        n (int, optional): number of instance for a sign. Defaults to 875.

    Returns:
        dict: A list associated to each sign
    """
    row = class_means.loc[label]
    return {
        sign: [row[f"{sign}_{i}"] for i in range(n)]
        for sign in [
            "EAR_left",
            "EAR_right",
            "EAR_mean",
            "MAR",
            "EBR",
            "PERCLOS",
            "HOP_gd",
            "HOP_hb"
        ]
    }
```

```python
class_means = data.groupby("class").mean(numeric_only=True)

# Mean of all instances signs
mean_signs_0 = extract_signs(class_means,0)
mean_signs_1 = extract_signs(class_means,1)
mean_signs_2 = extract_signs(class_means,2)
```

```python
import matplotlib.pyplot as plt

def plot_signs(signs_0, signs_1, signs_2):

    for sign in signs_0:
        plt.figure(figsize=(12, 5))

        plt.plot(signs_0[sign], label="Class 0")
        plt.plot(signs_1[sign], label="Class 1")
        plt.plot(signs_2[sign], label="Class 2")

        plt.title(sign)
        plt.xlabel("Frame")
        plt.ylabel(sign)
        plt.legend()
        plt.grid()
        plt.show()
```

```python
plot_signs(mean_signs_0, mean_signs_1, mean_signs_2)
```

### Distributions

Let's see the distributions of each sign, in order to interpret those distributions more easly i am not gonna plot the bins but the density curves of each sign.

```python
import numpy as np

signs = ["EAR_left", "EAR_right", "EAR_mean", "MAR", "EBR", "PERCLOS", "HOP_gd", "HOP_hb"]

# une colonne = moyenne du signe sur les 875 frames, pour chaque instance
instance_means = pd.DataFrame({
    sign: data[[f"{sign}_{i}" for i in range(875)]].mean(axis=1)
    for sign in signs
})

import seaborn as sns

for sign in signs:
    plt.figure(figsize=(10, 5))
    for c in sorted(data["class"].unique()):
        sns.kdeplot(instance_means[data["class"] == c][sign], label=f"Class {c}", linewidth=2)
    plt.title(sign)
    plt.xlabel(sign)
    plt.ylabel("Densité")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
```

For most of the signs we can see a different mean and std around it for each class, some of the fatigue signs are heavy tailed, specially HOP, EBR, PERCLOS and MAR. EAR is right skewed heavy tailed but not as much and still interpretable, they are closer to normal distribution

Let's transform those signs to see a better distribution

```python
def logit_transform(x, epsilon=1e-6):
    x_clipped = np.clip(x, epsilon, 1 - epsilon)
    return np.log(x_clipped / (1 - x_clipped))

signs_to_transform = ["EBR", "PERCLOS","HOP_hb", "MAR"] # HOP_gd is worst if we transform it 

instance_means_logit = instance_means.copy()
instance_means_logit[signs_to_transform] = instance_means[signs_to_transform].apply(logit_transform)
```

```python
for sign in signs:
    plt.figure(figsize=(10, 5))
    for c in sorted(data["class"].unique()):
        sns.kdeplot(instance_means_logit[data["class"] == c][sign], label=f"Class {c}", linewidth=2)
    plt.title(sign)
    plt.xlabel(sign)
    plt.ylabel("Densité")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
```

From what we can see from thoses plots, EAR is higher when the subject is more awaked and then decrease with the fatigue which is the complete opposite form MAR, HOP and EBR and PERCLOS, where we can see that the values are slightly higher but a lot more frequent than awake subjects.

### Correlations

Let's see how are the features correlated together, indeed some of them will be correlated according to their nature (e.g EAR, HOP)

```python
corr = instance_means.corr()

import seaborn as sns
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlations Between Signs")
plt.show()
```

The heatmap shows a few expected redundancies. EAR_left, EAR_right, and EAR_mean are highly correlated (0.85–0.97), as they all measure the same phenomenon from the left and right eyes. EBR and PERCLOS are also strongly correlated (0.78), since both are related to eye-closure behavior. HOP_gd and HOP_hb have a moderate correlation (0.51), as both are related to head orientation.

The most interesting relationships are the moderate negative correlations between EAR and EBR/PERCLOS (-0.39 to -0.63). This means that lower average eye openness is associated with more blinking and longer eye-closure periods, which is consistent with a fatigue-related signal. MAR has a weak to moderate positive correlation with EBR and PERCLOS (0.30–0.32), suggesting that yawning can sometimes occur together with blinking or eye closure, although this relationship is weaker than the one observed with EAR.

Finally, the correlations between MAR/EAR and the HOP features are weak (|r| < 0.21), so these relationships should not be over-interpreted based on this analysis alone.
